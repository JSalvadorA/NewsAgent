````python
import os
import shutil
import json
from datetime import datetime

# Función para limpiar un directorio específico
def clean_directory(directory_path, specific_dates=None, keep_dir=True):
    if os.path.exists(directory_path):
        # Si no hay fechas específicas, eliminamos todo el contenido
        if not specific_dates:
            for item in os.listdir(directory_path):
                item_path = os.path.join(directory_path, item)
                if os.path.isfile(item_path) or os.path.islink(item_path):
                    try:
                        os.unlink(item_path)
                        print(f"  - Eliminado: {item_path}")
                    except Exception as e:
                        print(f"  - Error al eliminar {item_path}: {e}")
                elif os.path.isdir(item_path):
                    try:
                        shutil.rmtree(item_path)
                        print(f"  - Eliminado: {item_path}")
                    except Exception as e:
                        print(f"  - Error al eliminar {item_path}: {e}")
            print(f"Contenido de '{os.path.basename(directory_path)}' eliminado.")
        else:
            # Eliminar solo archivos relacionados con las fechas específicas
            deleted_items = False
            for item in os.listdir(directory_path):
                # Verificar si el nombre del elemento contiene alguna de las fechas específicas
                if any(date in item for date in specific_dates):
                    item_path = os.path.join(directory_path, item)
                    if os.path.isfile(item_path) or os.path.islink(item_path):
                        try:
                            os.unlink(item_path)
                            print(f"  - Eliminado: {item_path}")
                            deleted_items = True
                        except Exception as e:
                            print(f"  - Error al eliminar {item_path}: {e}")
                    elif os.path.isdir(item_path):
                        try:
                            shutil.rmtree(item_path)
                            print(f"  - Eliminado: {item_path}")
                            deleted_items = True
                        except Exception as e:
                            print(f"  - Error al eliminar {item_path}: {e}")
            
            if deleted_items:
                print(f"Archivos de fechas específicas eliminados de '{os.path.basename(directory_path)}'.")
            else:
                print(f"No se encontraron archivos de las fechas especificadas en '{os.path.basename(directory_path)}'.")
        
        # Si no queremos mantener el directorio, lo eliminamos también
        if not keep_dir and not specific_dates:
            try:
                os.rmdir(directory_path)
                print(f"Directorio eliminado: {directory_path}")
                return
            except Exception as e:
                print(f"Error al eliminar directorio {directory_path}: {e}")
    else:
        print(f"El directorio '{os.path.basename(directory_path)}' no existe.")

# Ruta base del proyecto
project_root = r"C:/Jerson/SUNASS/2025/4_April/gem/scr1403"

# Solicitar fechas para eliminar archivos específicos
dates_input = input("Ingresa las fechas (DDMMYYYY) separadas por comas para eliminar archivos específicos (deja en blanco para eliminar todo): ").strip()

specific_dates = []
if dates_input:
    # Dividir la entrada por comas y eliminar espacios en blanco
    date_list = [date.strip() for date in dates_input.split(',')]
    
    # Validar cada fecha
    for date in date_list:
        try:
            datetime.strptime(date, '%d%m%Y')
            specific_dates.append(date)
            print(f"Fecha válida: {date}")
        except ValueError:
            print(f"Formato de fecha inválido: {date} - será ignorada")
    
    if specific_dates:
        print(f"\nEliminando archivos para las fechas: {', '.join(specific_dates)}")
    else:
        print("\nNo se ingresaron fechas válidas. Se limpiarán todos los archivos.")
else:
    print("\nEliminando todos los archivos generados.")

print("\n1. Limpiando la caché...")
# Limpiar la carpeta cache - siempre limpiamos toda la caché independientemente de las fechas
cache_folder = os.path.join(project_root, "cache")
clean_directory(cache_folder)

# Crear un nuevo archivo history.json con un objeto JSON vacío
history_file_path = os.path.join(cache_folder, "history.json")
with open(history_file_path, "w") as file:
    json.dump({}, file)
print("  - Archivo 'history.json' creado con contenido vacío.")

print("\n2. Limpiando archivos de salida...")

# Directorios a limpiar
directories_to_clean = [
    # Carpeta output principal
    os.path.join(project_root, "output"),
    
    # Input/Html
    os.path.join(project_root, "input", "Html"),
    
    # Input/Images/downloads
    os.path.join(project_root, "input", "Images", "downloads"),
    
    # Input/Out
    os.path.join(project_root, "input", "Out"),
    
    # Input/In
    os.path.join(project_root, "input", "In"),
    
    # Input/Social
    os.path.join(project_root, "input", "Social"),
    
    # Input/Stats
    os.path.join(project_root, "input", "Stats")
]

# Limpiar los directorios
for directory in directories_to_clean:
    print(f"\nRevisando {directory}...")
    if os.path.exists(directory):
        clean_directory(directory, specific_dates)
        
        # Para la carpeta de imágenes descargadas, revisar también las subcarpetas de fecha
        if directory == os.path.join(project_root, "input", "Images", "downloads") and specific_dates:
            for date in specific_dates:
                date_dir = os.path.join(directory, date)
                if os.path.exists(date_dir):
                    clean_directory(date_dir)
    else:
        print(f"  - Directorio {directory} no existe.")

# Limpiar la carpeta base
base_dir = os.path.join(project_root, "base")
if os.path.exists(base_dir):
    print(f"\nRevisando {base_dir}...")
    if specific_dates:
        # Si hay fechas específicas, solo limpiamos las subcarpetas correspondientes
        for date in specific_dates:
            date_dir = os.path.join(base_dir, date)
            if os.path.exists(date_dir):
                clean_directory(date_dir)
    else:
        # Si no hay fechas específicas, limpiamos toda la carpeta base
        clean_directory(base_dir)

# Limpiar archivos de logs
logs_dir = os.path.join(project_root, "logs")
if os.path.exists(logs_dir):
    print("\n3. Limpiando logs...")
    # Los logs se limpian completamente independientemente de las fechas
    clean_directory(logs_dir)

print("\n=== LIMPIEZA COMPLETA ===")
print("Sistema listo para procesar nuevamente archivos.")
print(f"Archivos eliminados para las fechas: {', '.join(specific_dates) if specific_dates else 'TODOS'}")
````

````python
import os
import shutil
import json
from datetime import datetime
import glob
import re

# Función para limpiar un directorio específico
def clean_directory(directory_path, specific_dates=None, keep_dir=True, pattern=None):
    if os.path.exists(directory_path):
        items_deleted = False
        
        # Obtener lista de archivos a eliminar
        items_to_check = os.listdir(directory_path)
        
        for item in items_to_check:
            item_path = os.path.join(directory_path, item)
            should_delete = False
            
            # Determinar si el archivo debe ser eliminado
            if specific_dates:
                # Si hay fechas específicas, verificar si el nombre contiene alguna de esas fechas
                if any(date in item for date in specific_dates):
                    should_delete = True
            elif pattern:
                # Si hay un patrón específico, verificar si el nombre coincide
                if re.search(pattern, item):
                    should_delete = True
            else:
                # Si no hay fechas ni patrón, eliminar todo
                should_delete = True
            
            # Eliminar el archivo/directorio si corresponde
            if should_delete:
                if os.path.isfile(item_path) or os.path.islink(item_path):
                    try:
                        os.unlink(item_path)
                        print(f"  - Eliminado: {item_path}")
                        items_deleted = True
                    except Exception as e:
                        print(f"  - Error al eliminar {item_path}: {e}")
                elif os.path.isdir(item_path):
                    try:
                        shutil.rmtree(item_path)
                        print(f"  - Eliminado: {item_path}")
                        items_deleted = True
                    except Exception as e:
                        print(f"  - Error al eliminar {item_path}: {e}")
        
        if not items_deleted:
            print(f"  - No se encontraron archivos para eliminar en '{os.path.basename(directory_path)}'")
        
        # Si no queremos mantener el directorio y está vacío, lo eliminamos
        if not keep_dir and not specific_dates and not os.listdir(directory_path):
            try:
                os.rmdir(directory_path)
                print(f"  - Directorio eliminado: {directory_path}")
                return
            except Exception as e:
                print(f"  - Error al eliminar directorio {directory_path}: {e}")
    else:
        print(f"  - El directorio '{os.path.basename(directory_path)}' no existe.")

# Función para buscar y eliminar archivos recursivamente según patrones
def clean_recursively(base_dir, specific_dates=None, file_patterns=None):
    for root, dirs, files in os.walk(base_dir):
        # Primero eliminamos archivos
        for file in files:
            file_path = os.path.join(root, file)
            should_delete = False
            
            # Verificar si el archivo coincide con alguna fecha específica
            if specific_dates and any(date in file for date in specific_dates):
                should_delete = True
            
            # Verificar si el archivo coincide con algún patrón
            elif file_patterns and any(re.search(pattern, file) for pattern in file_patterns):
                should_delete = True
            
            # Si no hay fechas ni patrones específicos, eliminar todo
            elif not specific_dates and not file_patterns:
                should_delete = True
                
            if should_delete:
                try:
                    os.unlink(file_path)
                    print(f"  - Eliminado: {file_path}")
                except Exception as e:
                    print(f"  - Error al eliminar {file_path}: {e}")
        
        # Luego intentamos eliminar directorios vacíos si no hay fechas específicas
        if not specific_dates:
            for dir_name in dirs[:]:  # Hacemos una copia para evitar problemas al modificar la lista mientras iteramos
                dir_path = os.path.join(root, dir_name)
                if not os.listdir(dir_path):  # Verificar si está vacío
                    try:
                        os.rmdir(dir_path)
                        print(f"  - Eliminado directorio vacío: {dir_path}")
                        dirs.remove(dir_name)  # Evitar entrar en directorios eliminados
                    except Exception as e:
                        print(f"  - Error al eliminar directorio vacío {dir_path}: {e}")

# Ruta base del proyecto
project_root = r"C:/Jerson/SUNASS/2025/4_April/gem/scr1403"

# Solicitar fechas para eliminar archivos específicos
dates_input = input("Ingresa las fechas (DDMMYYYY) separadas por comas para eliminar archivos específicos (deja en blanco para eliminar todo): ").strip()

specific_dates = []
if dates_input:
    # Dividir la entrada por comas y eliminar espacios en blanco
    date_list = [date.strip() for date in dates_input.split(',')]
    
    # Validar cada fecha
    for date in date_list:
        try:
            datetime.strptime(date, '%d%m%Y')
            specific_dates.append(date)
            print(f"Fecha válida: {date}")
        except ValueError:
            print(f"Formato de fecha inválido: {date} - será ignorada")
    
    if specific_dates:
        print(f"\nEliminando archivos para las fechas: {', '.join(specific_dates)}")
    else:
        print("\nNo se ingresaron fechas válidas. Se limpiarán todos los archivos.")
else:
    print("\nEliminando todos los archivos generados.")

print("\n1. Limpiando la caché...")
# Siempre limpiamos toda la caché independientemente de las fechas
cache_folder = os.path.join(project_root, "cache")
clean_directory(cache_folder)

# Crear un nuevo archivo history.json con un objeto JSON vacío
history_file_path = os.path.join(cache_folder, "history.json")
with open(history_file_path, "w") as file:
    json.dump({}, file)
print("  - Archivo 'history.json' creado con contenido vacío.")

# Patrones de archivos a buscar y eliminar (además de los que contienen la fecha)
file_patterns = [
    r'.*_unprocessed\.json$',
    r'.*\.csv$',
    r'texto_imagenes_api\.json$',
    r'consolidated_.*\.json$',
    r'scraped_texts_.*\.json$',
    r'facebook_.*\.json$',
    r'stats_.*\.json$',
    r'processing_stats_.*\.json$',
    r'extraction_results_.*\.json$',
    r'.*_results_.*\.json$',
    r'urls_.*\.json$',
    r'.*\.log$'
]

print("\n2. Limpiando archivos de salida recursivamente...")
# Limpieza recursiva de directorios principales
directories_to_clean_recursively = [
    os.path.join(project_root, "output"),
    os.path.join(project_root, "input"),
    os.path.join(project_root, "base")
]

for directory in directories_to_clean_recursively:
    if os.path.exists(directory):
        print(f"\nRevisando recursivamente {directory}...")
        clean_recursively(directory, specific_dates, file_patterns if specific_dates else None)
    else:
        print(f"  - El directorio '{directory}' no existe.")

# Limpiar subdirectorios específicos
print("\n3. Limpiando subdirectorios específicos...")
specific_subdirs = [
    os.path.join(project_root, "input", "Images", "downloads"),
    os.path.join(project_root, "input", "Html"),
    os.path.join(project_root, "input", "Out"),
    os.path.join(project_root, "input", "In"),
    os.path.join(project_root, "input", "Social"),
    os.path.join(project_root, "input", "Stats")
]

for directory in specific_subdirs:
    print(f"\nRevisando {directory}...")
    
    # Si hay fechas específicas, buscar directorios con esas fechas
    if specific_dates and os.path.exists(directory):
        for date in specific_dates:
            # Buscar subdirectorios con la fecha
            date_dirs = glob.glob(os.path.join(directory, f"*{date}*"))
            for date_dir in date_dirs:
                if os.path.isdir(date_dir):
                    print(f"  - Limpiando directorio de fecha: {date_dir}")
                    clean_directory(date_dir)
                else:
                    print(f"  - Eliminando archivo de fecha: {date_dir}")
                    try:
                        os.unlink(date_dir)
                    except Exception as e:
                        print(f"  - Error al eliminar {date_dir}: {e}")

# Limpiar archivos de logs
logs_dir = os.path.join(project_root, "logs")
if os.path.exists(logs_dir):
    print("\n4. Limpiando logs...")
    # Los logs se limpian completamente o según las fechas especificadas
    if specific_dates:
        for date in specific_dates:
            log_pattern = f".*{date}.*\.log$"
            for log_file in os.listdir(logs_dir):
                if re.search(log_pattern, log_file):
                    log_path = os.path.join(logs_dir, log_file)
                    try:
                        os.unlink(log_path)
                        print(f"  - Eliminado: {log_path}")
                    except Exception as e:
                        print(f"  - Error al eliminar {log_path}: {e}")
    else:
        clean_directory(logs_dir)

print("\n=== LIMPIEZA COMPLETA ===")
print("Sistema listo para procesar nuevamente archivos.")
print(f"Archivos eliminados para las fechas: {', '.join(specific_dates) if specific_dates else 'TODOS'}")
````

# Información del Proyecto NewsAg

## Descripción General

El proyecto es un sistema de procesamiento de documentos diseñado para extraer y analizar información de PDFs, imágenes, páginas web y publicaciones de Facebook. Está desarrollado para la SUNASS (Superintendencia Nacional de Servicios de Saneamiento) y permite automatizar la recopilación y análisis de contenido relacionado con servicios de agua y saneamiento en Perú.

## Estructura del Pipeline

El pipeline principal de procesamiento funciona de la siguiente manera:

1. **Extracción de enlaces del PDF**: Lee un PDF diario y extrae todos los enlaces.
2. **Extracción de texto del PDF**: Extrae el texto del PDF por secciones.
3. **Filtrado de URLs**: Compara con un historial para identificar solo URLs nuevas.
4. **Clasificación de URLs**: Categoriza los enlaces en imágenes, HTML, Facebook, etc.
5. **Procesamiento de imágenes**: Descarga las imágenes y extrae su texto mediante IA.
6. **Scraping de HTML**: Captura y analiza el contenido de páginas web.
7. **Procesamiento de audio**: Descarga y transcribe archivos de audio.
8. **Procesamiento de Facebook**: Captura publicaciones y genera PDFs.
9. **Extracción de texto de PDFs**: Utiliza OCR para PDFs escaneados.
10. **Consolidación de resultados**: Guarda todos los datos procesados en un archivo JSON unificado.

## Componentes Principales

El sistema se compone de varios módulos especializados:

- **ImageProcessor**: Descarga imágenes y extrae texto usando la API de Gemini AI.
- **HTMLScraper**: Extrae contenido de páginas web con relevancia para temas de agua.
- **FacebookProcessor**: Captura publicaciones de Facebook y las convierte a PDF.
- **HistoryTracker**: Mantiene un registro de URLs ya procesadas para evitar duplicados.
- **AudioProcessor**: Procesa y transcribe archivos de audio.
- **TextExtractor**: Extrae texto de PDFs, incluyendo OCR para documentos escaneados.

## Formas de Ejecución

El sistema ofrece tres formas diferentes de ejecución:

### 1. Notebook Interactivo (NewsAg.ipynb)

Diseñado para usuarios no técnicos, proporciona una interfaz gráfica con widgets que permite:
- Ingresar la fecha en formato DDMMYYYY
- Seleccionar el tipo de prompt (simple, detallado, estructurado, anti-ruido)
- Ajustar el tamaño de lote para procesamiento de imágenes
- Configurar el tiempo de pausa entre lotes
- Ejecutar el proceso con un botón

Este método fue específicamente desarrollado para capacitar a personas no familiarizadas con códigos ni comandos.

### 2. Script con Opciones (run_with_options.py)

Para ejecución desde terminal con parámetros personalizados:

```bash
python run_with_options.py DDMMYYYY [--batch-size N] [--pause-seconds N] [--prompt TIPO]
```

### 3. Ejecución Directa del Script Principal

Para usuarios avanzados:

```bash
python codigo/main.py DDMMYYYY
```

## Características Avanzadas

### Procesamiento Adaptativo de Imágenes

El sistema implementa un procesamiento adaptativo de imágenes que:
- Procesa imágenes en lotes secuenciales (tamaño configurable: 1-3)
- Reintenta automáticamente las imágenes que fallan en el primer intento
- Continúa con las demás imágenes si una falla después del reintento
- Permite configurar el tamaño del lote y la pausa entre procesamiento

### Manejo de Multiples Fechas

Mediante el script `run_multi_dates.py`, el sistema puede procesar:
- Una fecha específica
- Un rango de fechas 
- Los últimos N días

### Procesamiento específico de Facebook

Con el script `process_facebook.py` se puede procesar exclusivamente contenido de Facebook para una fecha determinada.

## Salidas del Sistema

Al ejecutar el sistema, se generan los siguientes resultados:

1. **Texto extraído de imágenes**: Archivos JSON con texto reconocido en imágenes.
2. **Contenido web extraído**: Datos estructurados de páginas web relevantes.
3. **PDFs de publicaciones de Facebook**: Capturas de publicaciones convertidas a PDF.
4. **Transcripciones de audio**: Texto extraído de archivos de audio.
5. **Texto de PDFs escaneados**: Texto extraído con OCR de documentos escaneados.
6. **Archivo consolidado**: Un JSON que integra todos los datos procesados para una fecha específica.

## Recomendaciones de Uso

- Para imágenes ligeras: batch_size=3, pause_seconds=60
- Para imágenes más complejas: batch_size=2, pause_seconds=60-90
- Para imágenes muy pesadas: batch_size=1, pause_seconds=90-120

El notebook `NewsAg.ipynb` es la forma recomendada para usuarios no técnicos, ya que proporciona una interfaz fácil de usar con controles intuitivos y no requiere conocimientos de programación o línea de comandos.

````python
import google.generativeai as genai
from PIL import Image
import os
import argparse
from dotenv import load_dotenv # Importar la función para cargar .env

# --- Definición de Prompts Predefinidos ---
# Aquí puedes añadir o modificar prompts según tus pruebas y necesidades.
# Las claves ('simple', 'detallado', etc.) se usarán para seleccionar el prompt.
PREDEFINED_PROMPTS = {
    "simple": (
        "Extrae únicamente el texto legible que encuentres en esta imagen de documento escaneado. "
        "Ignora por completo cualquier logotipo, gráfico, figura, fotografía, diagrama o elemento visual similar. "
        "No describas la imagen, solo transcribe el texto."
    ),
    "detallado": (
        "Realiza una transcripción OCR del contenido textual de esta imagen. "
        "Tu objetivo es capturar todo el texto escrito, incluyendo titulares, párrafos, pies de foto (solo el texto), etc. "
        "Excluye explícitamente cualquier elemento no textual como imágenes, gráficos, bordes decorativos, y logotipos. "
        "No interpretes ni resumas, solo transcribe."
    ),
    "estructurado": (
        "Analiza la estructura de esta página (probablemente un diario o documento similar) y extrae todo el texto de los artículos, titulares y bloques de texto. "
        "Omite deliberadamente cualquier imagen, publicidad gráfica, gráfico estadístico, o logotipo. "
        "Conserva los saltos de párrafo si es posible, pero enfócate en obtener solo el contenido escrito."
    ),
    "anti-ruido": (
        "Transcribe el texto principal de este documento. Presta especial atención a ignorar elementos visuales distractores como manchas, sellos superpuestos (si no son texto claro), firmas (si son ilegibles o puramente gráficas), y cualquier tipo de gráfico o ilustración. "
        "Devuelve solo el texto puro."
    )
}

# --- Configuración (Carga de API Key desde .env) ---
script_dir = os.path.dirname(os.path.abspath(__file__))
dotenv_path = os.path.join(script_dir, 'credentials', '.env')

if os.path.exists(dotenv_path):
    load_dotenv(dotenv_path=dotenv_path)
    print(f"Archivo .env cargado desde: {dotenv_path}")
else:
    print(f"Advertencia: No se encontró el archivo .env en: {dotenv_path}. Asegúrate de que exista o que la variable de entorno GOOGLE_API_KEY esté definida globalmente.")

api_key = os.getenv("GOOGLE_API_KEY")

if not api_key:
    raise ValueError("No se encontró la variable de entorno 'GOOGLE_API_KEY'. "
                     "Asegúrate de que esté definida en el archivo 'credentials/.env' o globalmente.")

try:
    genai.configure(api_key=api_key)
    print("API Key de Gemini configurada exitosamente.")
except Exception as e:
    print(f"Error al configurar la API Key de Gemini: {e}")
    # exit() # Considera salir si la configuración falla

# --- Función para extraer texto (sin cambios internos) ---
def extract_text_from_image(image_path: str, prompt: str) -> str:
    """
    Extrae texto de una imagen usando la API de Gemini (ya configurada)
    con el prompt proporcionado.

    Args:
        image_path (str): Ruta al archivo de imagen.
        prompt (str): La instrucción específica para el modelo Gemini.

    Returns:
        str: El texto extraído por el modelo.
        None: Si ocurre un error.
    """
    try:
        if not os.path.exists(image_path):
            print(f"Error: El archivo de imagen no se encontró en: {image_path}")
            return None

        img = Image.open(image_path)
        # Modelo multimodal (puedes probar 'gemini-1.5-flash-latest' para más velocidad/menos costo)
        model = genai.GenerativeModel('gemini-1.5-pro-latest')

        print(f"Procesando imagen: {image_path}...")
        print(f"Usando prompt: \"{prompt[:100]}...\"") # Muestra inicio del prompt usado

        response = model.generate_content([prompt, img])

        # print("\nRespuesta completa de la API (para depuración):")
        # print(response)
        # print(f"\nPrompt Feedback: {response.prompt_feedback}")

        if response.parts:
             return response.text
        else:
            print("Advertencia: La respuesta del modelo no contiene partes de texto.")
            if response.prompt_feedback:
                print(f"Feedback del prompt (puede indicar bloqueo o problema): {response.prompt_feedback}")
            return None

    except FileNotFoundError:
        print(f"Error: Archivo no encontrado en la ruta: {image_path}")
        return None
    except Exception as e:
        print(f"Ocurrió un error al procesar la imagen {image_path}: {e}")
        # import traceback
        # print(traceback.format_exc())
        return None

# --- Cómo usar la función ---
if __name__ == "__main__":
    parser = argparse.ArgumentParser(description="Extraer texto de una imagen usando Gemini API con prompts predefinidos.")
    parser.add_argument("image_path", help="Ruta al archivo de imagen.")
    # Nuevo argumento para seleccionar el prompt predefinido
    parser.add_argument(
        "-pk", "--prompt_key",
        choices=PREDEFINED_PROMPTS.keys(), # Las opciones son las claves del diccionario
        default='simple',                 # Prompt por defecto si no se especifica
        help="Clave del prompt predefinido a usar para la extracción de texto. Opciones: " + ", ".join(PREDEFINED_PROMPTS.keys())
    )
    # Argumento opcional para un prompt totalmente personalizado (si se usa, ignora -pk)
    parser.add_argument(
        "-cp", "--custom_prompt",
        type=str,
        default=None,
        help="Proporciona un prompt personalizado en lugar de usar uno predefinido."
    )


    args = parser.parse_args()

    # Determinar qué prompt usar
    if args.custom_prompt:
        selected_prompt = args.custom_prompt
        print(f"Usando prompt personalizado: \"{selected_prompt[:100]}...\"")
    else:
        selected_prompt = PREDEFINED_PROMPTS[args.prompt_key]
        print(f"Usando prompt predefinido (clave: '{args.prompt_key}')")
        # No es necesario imprimir el prompt completo aquí, ya se hace en la función

    # Llama a la función principal con la imagen y el prompt seleccionado
    extracted_text = extract_text_from_image(args.image_path, selected_prompt)

    if extracted_text:
        print("\n--- Texto Extraído ---")
        print(extracted_text)

        img_dir, img_filename = os.path.split(args.image_path)
        base_filename = os.path.splitext(img_filename)[0]
        # Añadir la clave del prompt usado al nombre del archivo de salida (si no es personalizado)
        prompt_suffix = f"_prompt_{args.prompt_key}" if not args.custom_prompt else "_prompt_custom"
        output_filename = os.path.join(img_dir, f"{base_filename}{prompt_suffix}_texto_extraido.txt")

        try:
            with open(output_filename, "w", encoding="utf-8") as f:
                f.write(extracted_text)
            print(f"\nTexto guardado en: {output_filename}")
        except Exception as e:
            print(f"Error al guardar el archivo de texto: {e}")
    else:
        print("\nNo se pudo extraer texto de la imagen o la respuesta estaba vacía.")
````

## Explicación de los ejemplos de curl

Los ejemplos de curl que mencionas son comandos para interactuar con el servidor Puppeteer MCP:

1. **Launch browser**: 
   ```bash
   curl -X POST http://localhost:3000/launch -H "Content-Type: application/json" -d '{"id":"main"}'
   ```
   Este comando inicia una nueva instancia de navegador con el ID "main". El servidor mantiene un registro de esta instancia para operaciones futuras.

2. **Navigate to a URL**:
   ```bash
   curl -X POST http://localhost:3000/navigate -H "Content-Type: application/json" -d '{"id":"main","url":"https://example.com"}'
   ```
   Este comando hace que el navegador con ID "main" navegue a la URL https://example.com.

3. **Take a screenshot**:
   ```bash
   curl -X POST http://localhost:3000/screenshot -H "Content-Type: application/json" -d '{"id":"main","path":"screenshot.png"}'
   ```
   Este comando toma una captura de pantalla de la página actual en el navegador "main" y la guarda como "screenshot.png".

4. **Close browser**:
   ```bash
   curl -X POST http://localhost:3000/close -H "Content-Type: application/json" -d '{"id":"main"}'
   ```
   Este comando cierra el navegador con ID "main" y libera los recursos asociados.

Estos comandos son ejemplos de cómo interactuar con el servidor desde la línea de comandos, pero también puedes usar cualquier lenguaje de programación que pueda hacer solicitudes HTTP, como se muestra en el archivo example-client.js con JavaScript.
